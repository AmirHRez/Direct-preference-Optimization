```python
from datasets import load_dataset
```

```python
dataset = load_dataset("yahma/alpaca-cleaned", split="train")
```

```python
dataset = dataset.filter(
    lambda x: x["input"] is None or x["input"].strip() == ""
)
len(dataset)
```

```python
def add_word_count(example):
    instruction_words = len(example["instruction"].split())
    output_words = len(example["output"].split())

    example["instruction_words"] = instruction_words
    example["output_words"] = output_words
    example["total_words"] = instruction_words + output_words

    return example
```

```python
dataset = dataset.map(add_word_count)
```

```python
dataset = dataset.filter(
    lambda x: x["total_words"] <= 200
)
len(dataset)
```

```python
dataset = dataset.shuffle(seed=42).select(range(3500))
```

```python
dataset.save_to_disk("alpaca_3500")
```
